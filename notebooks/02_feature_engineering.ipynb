```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/credit_risk_cleaned.csv")
df.head()
```

```python
df["AGE_GROUP"] = pd.cut(
    df["AGE"],
    bins=[0, 30, 40, 50, 60, 100],
    labels=[1, 2, 3, 4, 5]
)
```

```python
bill_cols = ["BILL_AMT1","BILL_AMT2","BILL_AMT3","BILL_AMT4","BILL_AMT5","BILL_AMT6"]
df["TOTAL_BILL"] = df[bill_cols].sum(axis=1)
df["AVG_BILL"] = df[bill_cols].mean(axis=1)
```

```python
pay_amt_cols = ["PAY_AMT1","PAY_AMT2","PAY_AMT3","PAY_AMT4","PAY_AMT5","PAY_AMT6"]
df["TOTAL_PAYMENT"] = df[pay_amt_cols].sum(axis=1)
df["AVG_PAYMENT"] = df[pay_amt_cols].mean(axis=1)
```

```python
df["CREDIT_UTILIZATION"] = (
    df["TOTAL_BILL"] /
    (df["LIMIT_BAL"] + 1))  #very important feature to distinguish two customers who have same due but different limits
```

```python
df["PAYMENT_RATIO"] = (
    df["TOTAL_PAYMENT"] /
    (df["TOTAL_BILL"].abs() + 1))
```

```python
df[
    [
        "AGE_GROUP",
        "TOTAL_BILL",
        "AVG_BILL",
        "TOTAL_PAYMENT",
        "AVG_PAYMENT",
        "CREDIT_UTILIZATION",
        "PAYMENT_RATIO"
    ]
].head()
```

```python
pay_status_cols = ["PAY_0","PAY_2","PAY_3","PAY_4","PAY_5","PAY_6"]

df["AVG_DELAY"] = df[pay_status_cols].mean(axis=1)

df["MAX_DELAY"] = df[pay_status_cols].max(axis=1)

df["NUM_DELAY_MONTHS"] = (df[pay_status_cols] > 0).sum(axis=1)
```

```python
df["RECENT_DELAY_FLAG"] = (df["PAY_0"] > 0).astype(int)
#did the customer recently missed a payment?
```

```python
df["BILL_GROWTH"] = (df["BILL_AMT1"] - df["BILL_AMT6"])
#Large positive growth may indicate worsening financial behaviour
```

```python
df["PAYMENT_STD"] = (df[pay_amt_cols].std(axis=1))
#we did this to check how consistently the customer is paying his bills
```

```python
df[
    [
        "AVG_DELAY",
        "MAX_DELAY",
        "NUM_DELAY_MONTHS",
        "RECENT_DELAY_FLAG",
        "BILL_GROWTH",
        "PAYMENT_STD"
    ]
].head()
```
