```python
import pandas as pd

pd.set_option("display.max_columns", None)
df = pd.read_csv(
    "../data/raw/default of credit card clients.csv",
    header=1
)

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.info()
```

```python
df.duplicated().sum()
```

```python
df.describe().T #to see if any value looks strange, not much required here 
```

```python
print("EDUCATION Values:")
print(sorted(df["EDUCATION"].unique()))

print("\nMARRIAGE Values:")
print(sorted(df["MARRIAGE"].unique()))
```

```python
print(df["EDUCATION"].value_counts().sort_index())
print(df["MARRIAGE"].value_counts().sort_index())
```

```python
for col in ["EDUCATION", "MARRIAGE"]:
    print(f"\n{col}")
    print(df[col].value_counts(normalize=True) * 100)
    #normalize=True, computes proportions instead of counts and then we calc the percentage
```

```python
df["EDUCATION"] = df["EDUCATION"].replace([0, 5, 6],4)
df["MARRIAGE"] = df["MARRIAGE"].replace(0,3)
```

```python
pay_cols = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

for col in pay_cols:
    print(f"\n{col}")
    print(sorted(df[col].unique()))
```

```python
bill_cols = [
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6"
]
for col in bill_cols:
    count_neg = (df[col] < 0).sum()

    print(f"{col}: {count_neg} negative values ")
```

```python
pay_amt_cols = [
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]
for col in pay_amt_cols:
    count_neg = (df[col] < 0).sum()

    print(f"{col}: {count_neg} negative values ")
```

```python
df.to_csv(
    "../data/processed/credit_risk_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
